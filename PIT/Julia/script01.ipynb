# Qué es Julia?

Julia es un lenguaje de programaición dinámico, de alto nivel y de alto
rendimiento, diseñado específicamente para computación técnica y ciéntifica.

**Objetivo**: Resolver el problema de los dos lenguajes

> Sintaxis amigable similar a Python o MATLAB.

> Velocidad comparable a la de C o Fortran.

> Diseñadod desde cero para el paralelismo y la computación en la nube.

# Características clave de Julia

* Rápido: Compilado Just-in-Time (JIT) a través de LLVM.
* Dinámico: Tipado dinámico por defecto.
* Reproducible: Pkg (Gestor de Paquetes) asegura entornos de proyectos consistentes y reproducibles.
* Composable: El despacho múlitple permite que paquetes desarrollados de manera independiente funcionen juntos sin problemas
* General: Aunque esté optimizado para la ciencia, hoy en día es un lenguaje de propósito general capaz de desarrollar aplicaciones web (backend) y microservicios.
* Interoperable: Puede llamar directamente a librerias de C, Fortran, Python(PyCall.jl) y R (RCall.jl).


# Ecosistema y Comunidad
* Gestor de paquetes (Pkg.jl): Paquete integrado en el lenguaje, maneja dependencias a nivel de proyecto para lograr reproducibilidad perfecta.
