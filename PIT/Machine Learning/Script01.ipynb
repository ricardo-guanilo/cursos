EL modulo `matplotlib`

```python
# Ejemplito1
x = [0.2,0.8,1.2,-1.2,-3.6,1.3]
y = [1,2,4,6,8,9]

# Diagrama  de dispersion
plt.plot(x,y)

# fin
plt.show()
```

```python
# Ejemplito2
# La funcion linspace
# help(np.linspace)
x1 = np.linspace(-5,5,100)
y1 = np.sin(x1**(1/2))


# Diagrama de dispersion
plt.plot(x1,y1)

# fin
plt.show()
```

```python
# Ejemplito3
# Grafiquemos sin(x)/x en [-0.5,0.5]
x2 = np.linspace(-0.5,0.5,100)
y2 = np.sin(x2)/x2

# Diagrama de dispersion
plt.plot(x2,y2)

# fin
plt.show()
```

```python
# Ejemplito4
# Consideremos otra regla de correspondencia : y = sin(x**2)*exp(x**2) en [4.5,5.5]
x3 = np.linspace(4.5, 5.5, 100)
y3 = np.sin(x3**2)*np.exp(x3**2)

# Mostremos el diagrama de dispersion
plt.plot(x3,y3)

# fin
plt.show()
```

```python
# Ejemplito5 : COnsideremos colocar dos reglas de correspondencia

# Definamos la variable independiente
x4 = np.linspace(2,18,120)

# Definamos las dos reglas de correspondencia a mostrar
y4_A = np.sin(x4**(1/7))
y4_B = np.exp(-x4**(2))

# En el mismo grafico observar las dos reglas de correspondencia
plt.plot(x4,y4_A,"r" ,x4,y4_B + np.random.normal(0,0.5,size = len(y4_B))*10**(-2),"b")

# fin
plt.show()
```

```python
# Ejemplito6 : Mostremos cada señal en una ventana

# ventana1
plt.figure(num = 1)
plt.plot(x4,y4_A,"r", label = "Fun. Trigonometrica")
# Especificar el intervalo para el eje-Y
plt.ylim(-0.1,1.2)
# Agreguemos etiquetas
plt.title("Funcion y4_A", {"fontsize" : 24, "color" : "red"})
plt.xlabel("Variable Independiente")
plt.ylabel("Producto1")
# Agreguemos la etiqueta configurada en plt.plot
plt.legend()

# ventan2
plt.figure(num = 2)
plt.plot(x4,y4_B + np.random.normal(0,0.5,size = len(y4_B))*10**(-2),"b", label = "Fun. Exponencial")
# Especificar el intervalo para el eje-Y
plt.ylim(-0.1,1.2)
# Agreguemos etiquetas
plt.title("Funcion y4_B + Ruido", {"fontsize" : 24, "color" : "blue"})
plt.xlabel("Variable Independiente")
plt.ylabel("Producto2")
# Agreguemos la etiqueta configurada en plt.plot
plt.legend()


# fin
plt.show()
```
